### Security of local files
Bench copy 21.


Note that this means that the Jupyter notebook server also acts as a generic file server

Bench copy 22.
for files inside the same tree as your notebooks.  Access is not granted outside the

Bench copy 22.
notebook folder so you have strict control over what files are visible, but for this

Bench copy 22.
reason it is highly recommended that you do not run the notebook server with a notebook

Bench copy 22.
directory at a high level in your filesystem (e.g. your home directory).

Bench copy 22.


Bench copy 22.
When you run the notebook in a password-protected manner, local file access is restricted

Bench copy 22.
to authenticated users unless read-only views are active.
Bench copy 22.

### Security of local files
Bench copy 1341.


Note that this means that the Jupyter notebook server also acts as a generic file server

Bench copy 1342.
for files inside the same tree as your notebooks.  Access is not granted outside the

Bench copy 1342.
notebook folder so you have strict control over what files are visible, but for this

Bench copy 1342.
reason it is highly recommended that you do not run the notebook server with a notebook

Bench copy 1342.
directory at a high level in your filesystem (e.g. your home directory).

Bench copy 1342.


Bench copy 1342.
When you run the notebook in a password-protected manner, local file access is restricted

Bench copy 1342.
to authenticated users unless read-only views are active.

<https://daringfireball.net/projects/markdown/>

### Security of local files
Bench copy 2205.


Note that this means that the Jupyter notebook server also acts as a generic file server

Bench copy 2206.
for files inside the same tree as your notebooks.  Access is not granted outside the

Bench copy 2206.
notebook folder so you have strict control over what files are visible, but for this

Bench copy 2206.
reason it is highly recommended that you do not run the notebook server with a notebook

Bench copy 2206.
directory at a high level in your filesystem (e.g. your home directory).

Bench copy 2206.


Bench copy 2206.
When you run the notebook in a password-protected manner, local file access is restricted

Bench copy 2206.
to authenticated users unless read-only views are active.
Bench copy 2206.

<https://daringfireball.net/projects/markdown/>

### Security of local files
Bench copy 3669.


Note that this means that the Jupyter notebook server also acts as a generic file server

Bench copy 3670.
for files inside the same tree as your notebooks.  Access is not granted outside the

Bench copy 3670.
notebook folder so you have strict control over what files are visible, but for this

Bench copy 3670.
reason it is highly recommended that you do not run the notebook server with a notebook

Bench copy 3670.
directory at a high level in your filesystem (e.g. your home directory).

Bench copy 3670.


Bench copy 3670.
When you run the notebook in a password-protected manner, local file access is restricted

Bench copy 3670.
to authenticated users unless read-only views are active.
Bench copy 3670.

<https://daringfireball.net/projects/markdown/>

### Security of local files
Bench copy 5133.


Note that this means that the Jupyter notebook server also acts as a generic file server

Bench copy 5134.
for files inside the same tree as your notebooks.  Access is not granted outside the

Bench copy 5134.
notebook folder so you have strict control over what files are visible, but for this

Bench copy 5134.
reason it is highly recommended that you do not run the notebook server with a notebook

Bench copy 5134.
directory at a high level in your filesystem (e.g. your home directory).

Bench copy 5134.


Bench copy 5134.
When you run the notebook in a password-protected manner, local file access is restricted

Bench copy 5134.
to authenticated users unless read-only views are active.
Bench copy 5134.

### Security of local files
Bench copy 6597.


Note that this means that the Jupyter notebook server also acts as a generic file server

Bench copy 6598.
for files inside the same tree as your notebooks.  Access is not granted outside the

Bench copy 6598.
notebook folder so you have strict control over what files are visible, but for this

Bench copy 6598.
reason it is highly recommended that you do not run the notebook server with a notebook

Bench copy 6598.
directory at a high level in your filesystem (e.g. your home directory).

Bench copy 6598.


Bench copy 6598.
When you run the notebook in a password-protected manner, local file access is restricted

Bench copy 6598.
to authenticated users unless read-only views are active.
Bench copy 6598.
